```python
infer = YOLO('/content/runs/detect/train/weights/best.pt')
```

```python
# Required libraries
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt

# Load image from the URL
url = 'https://s3.amazonaws.com/plantvillage-production-new/images/pics/000/000/626/original/Early_blight_3.jpg?1369844363'
response = requests.get(url)
img = Image.open(BytesIO(response.content))

# Run inference on the image (without saving it)
results = infer.predict(source=img, save=False, save_txt=False)

# Display the image and overlay predictions using matplotlib
plt.imshow(results[0].plot())  # Use YOLO's plot method to visualize predictions
plt.axis('off')  # Hide axes
plt.show()
```
